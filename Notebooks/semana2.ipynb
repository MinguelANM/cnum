 ## Atividade 1
Faça na linguagem Python uma função que verifica se um númeor é perfeito.
 
Um número é perfeito se a soma dos divisores for igual a ele mesmo (ex: 6 = 1 +2 +3).

````python
def es_perfeito(numero: int) -> bool:
    if numero <= 1:
        return False
    
    soma_divisores = 0
    for i in range(1, (numero // 2) + 1):
        if numero % i == 0:
            soma_divisores += i
    return soma_divisores == numero

def main():
    numero = 6
    if es_perfeito(numero):
        print(f"O número {numero} é perfeito!")
    else:
        print(f"O número {numero} não é perfeito.")


if __name__ == "__main__":
    main()
````

 ## Atividade 2
Faça na linguagem Python uma função que calcula o fatorial de um número.
 

````python
def fatorial(n: int) -> int:
    if n < 0:
        raise ValueError("Não existe fatorial de número negativo.")
    
    resultado = 1
    for i in range(1, n + 1):
        resultado *= i
        
    return resultado


def main():
    numero = 5
    print(f"O fatorial de {numero} é {fatorial(numero)}")


if __name__ == "__main__":
    main()
````
