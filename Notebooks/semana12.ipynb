## Atividade 1
Escreva os métodos regra de quadratura pelo ponto médio, Trapézio e Simpson dentro do arquivo $utils/algoritmos.py$.

````python
import sys
import os
import numpy as np
import matplotlib.pyplot as plt

sys.path.append(os.path.abspath(os.path.join('..')))
from scipy.integrate import quad

from utils.algoritmos import medio, trapezio, simpson, integral
````

Calcule numericamente as seguintes integrais:

a) $\displaystyle \int_{0}^{1} e^{-x} \, dx$  
b) $\displaystyle \int_{0}^{1} x^2 \, dx$  
c) $\displaystyle \int_{0}^{1} x^3 \, dx$  
d) $\displaystyle \int_{0}^{1} x e^{-x^2} \, dx$  
e) $\displaystyle \int_{0}^{1} \frac{1}{x^2 + 1} \, dx$  
f) $\displaystyle \int_{0}^{1} \frac{x}{x^2 + 1} \, dx$

Use os métodos simples do **ponto médio**, **Trapézio** e **Simpson**.  

**Respostas esperadas:**

| $f$ | Ponto médio | Trapézio | Simpson | SciPy | 
|:--:|:--:|:--:|:--:|:--:|
| a | 0.63175246 | 0.63175254 | 0.63175249 | 0.63212055
| b | 0.33233425 | 0.33233450 | 0.33233433 | 0.33333333
| c | 0.24900137 | 0.24900175 | 0.24900150 | 0.25000000
| d | 0.31569227 | 0.31569210 | 0.31569222 | 0.31606028
| e | 0.78489793 | 0.78489787 | 0.78489791 | 0.78539816
| f | 0.34607363 | 0.34607351 | 0.34607359 | 0.34657359

````python
funcoes = {
    "a": lambda x: np.exp(-x),
    "b": lambda x: x**2,
    "c": lambda x: x**3,
    "d": lambda x: x * np.exp(-x**2),
    "e": lambda x: 1 / (x**2 + 1),
    "f": lambda x: x / (x**2 + 1),
}

n = 1000
print("f | Ponto médio | Trapézio | Simpson | SciPy")
for nome, f in funcoes.items():
    r = [integral(m, f, 0, 1, n) for m in (medio, trapezio, simpson)]
    s, _ = quad(f, 0, 1)
    print(f"{nome} | {r[0]:.8f} | {r[1]:.8f} | {r[2]:.8f} | {s:.8f}")
````

## Atividade 2
Calcule numericamente o valor de  

$\displaystyle \int_{2}^{5} e^{4 - x^2} \, dx$  

usando os métodos **compostos** do **ponto médio**, **Trapézio** e **Simpson**.  
Obtenha os resultados utilizando, em cada quadratura, o número de pontos indicado.

**Respostas esperadas:**

| $n$ | Ponto médio | Trapézios | Simpson |
|:--:|:--:|:--:|:--:|
| 3 | 0.1056606 | 0.7503919 | 0.5005225 |
| 5 | 0.1726140 | 0.3964724 | 0.2784992 |
| 7 | 0.1973663 | 0.3062023 | 0.2393551 |
| 9 | 0.2084204 | 0.2721145 | 0.2306618 |

````python
f = lambda x: np.exp(4 - x**2)

print("n | Ponto médio | Trapézios | Simpson")
for n in (3, 5, 7, 9):
    pm = integral(medio, f, 2, 5, n)
    tr = integral(trapezio, f, 2, 5, n - 1)
    si = integral(simpson, f, 2, 5, (n - 1) // 2)
    print(f"{n} | {pm:.7f} | {tr:.7f} | {si:.7f}")
````
