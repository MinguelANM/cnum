## Atividade 1
Escreva os métodos de ponto fixo dentro do arquivo $utils/algoritmos.py$.

````python
import numpy as np
from scipy.optimize import approx_fprime

def pontofixo(a, g, TOL=1e-8):
    x = g(a)
    while abs(x - a) > TOL:
        a = x
        x = g(a)
    return x

def newton_raphson(a, f, TOL=1e-8, df=None):
    if df is None:
        def dfn(x):
            return approx_fprime(np.array([x]), lambda v: f(v[0]))[0]
    else:
        dfn = df
    g = lambda x: x - f(x) / dfn(x)
    return pontofixo(a, g, TOL)

def secante(a, b, f, TOL=1e-8):
    g = lambda a, b: (a * f(b) - b * f(a)) / (f(b) - f(a))
    x = g(a, b)
    while abs(x - b) > TOL:
        a, b = b, x
        x = g(a, b)
    return x
````

Resolver a equação $e^x = x + 2$ é equivalente a calcular os pontos fixos da função  

$$
g(x) = e^x - 2
$$

Use os métodos do ponto fixo  

$$
x^{(n+1)} = g(x^{(n)})
$$

com $x^{(0)} = -1.8$ para obter uma aproximação de uma das soluções da equação dada com 8 dígitos significativos.  

**Resposta:**  
$
x \approx -1.8414057
$

````python
import numpy as np

import sys
import os

# Sobe um nível para a raiz do projeto e adiciona ao caminho do sistema
sys.path.append(os.path.abspath(os.path.join('..')))

from utils.algoritmos import (
    pontofixo,
    newton_raphson,
    secante,
)

f1 = lambda x: np.e**x - x - 2
g1 = lambda x: np.e**x - 2


def main():
    r = pontofixo(-1.8, g1)
    print(f"raiz ponto fixo = {r}")
    r = newton_raphson(-1.8, f1, df=lambda x: np.e**x - 1)
    print(f"raiz newton-raphson df = {r}")
    r = newton_raphson(-1.8, f1)
    print(f"raiz newton-raphson = {r}")
    r = secante(-1.8, -1.7, f1)
    print(f"raiz secante = {r}")


if __name__ == "__main__":
    main()
````

## Atividade 2
Encontre a raiz positiva da função  

$$
f(x) = \cos(x) - x^2
$$  

pelos métodos do ponto fixo, inicializando-o com $x^{(0)} = 1$.  

Realize a iteração até obter estabilidade no **quinto dígito significativo**.  

**Resposta:**

$$
x \approx 0.82413 
$$

Processo iterativo:  

$$
x^{(n+1)} = x^{(n)} + \frac{\cos(x) - x^2}{\sin(x) + 2x}
$$

````python
import numpy as np
from scipy.optimize import approx_fprime

def pontofixo(a, g, TOL=1e-5):
    x = g(a)
    while abs(x - a) > TOL:
        a = x
        x = g(a)
    return x

def newton_raphson(a, f, TOL=1e-5, df=None):
    if df is None:
        def dfn(x):
            return approx_fprime(np.array([x]), lambda v: f(v[0]))[0]
    else:
        dfn = df
    g = lambda x: x - f(x) / dfn(x)
    return pontofixo(a, g, TOL)

# 1. Definindo a função e sua derivada exata
f = lambda x: np.cos(x) - x**2
df = lambda x: -np.sin(x) - 2*x

# 2. Executando com Newton-Raphson (que usa a função pontofixo internamente)
x0 = 1.0
raiz = newton_raphson(x0, f, TOL=1e-5, df=df)

print(f"Raiz encontrada: {raiz:.5f}")
````

## Atividade 3
Aplique os métodos do ponto fixo para resolver a equação:

$$
e^{-x^2} = 2x
$$

**Resposta:**

$$
x \approx 0.4193648
$$

````python
import numpy as np

def pontofixo(a, g, TOL=1e-8):
    x = g(a)
    while abs(x - a) > TOL:
        a = x
        x = g(a)
    return x

# Definindo a função de iteração g(x)
g = lambda x: np.exp(-x**2) / 2

# Chute inicial (pode ser x0 = 0 ou 0.5)
x0 = 0.5

# Execução
raiz = pontofixo(x0, g, TOL=1e-8)

print(f"Raiz encontrada: {raiz:.7f}")
````

### Atividade 5

````python
import numpy as np

def pontofixo(a, g, TOL=1e-3):
    x = g(a)
    while abs(x - a) > TOL:
        a = x
        x = g(a)
    return x

vt = 0.025852
Ir = 1e-12

casos = [
    (30.0, 1000.0),    # a
    (3.0, 1000.0),     # b
    (3.0, 10000.0),    # c
    (0.3, 1000.0),     # d
    (-0.3, 1000.0),    # e
    (-30.0, 1000.0),   # f
    (-30.0, 10000.0)   # g
]

for V, R in casos:
    # Para V positivo, isolamos vd do expoente: vd = vt * ln(1 + (V - vd)/(R*Ir))
    # Para V negativo, Id é insignificante, vd aproxima V
    if V > 0:
        g = lambda vd: vt * np.log(1 + max(0.0, V - vd) / (R * Ir))
        v0 = 0.5
    else:
        g = lambda vd: V - R * Ir * (np.exp(vd / vt) - 1)
        v0 = V

    res = pontofixo(v0, g, TOL=1e-3)
    print(f"V = {V:5.1f}V, R = {R:7.1f} \u03a9 -> vd \u2248 {res:.3f} V")
````
